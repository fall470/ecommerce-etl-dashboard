```python
import pandas as pd

#Charger les données brutes
df = pd.read_csv("data/ecommerce.csv")

print("Aperçu du dataset brut :")
display(df.head())
display(df.info())
display(df.describe(include="all"))
```

```python
# Gestion des valeurs manquantes

print("\nValeurs manquantes avant nettoyage :")
print(df.isnull().sum())
```

```python
# Remplacer NaN dans Returns par 0
df["Returns"].fillna(0, inplace=True)
```

```python
# Suppression des colonnes inutiles
cols_to_drop = ["Customer Name", "Age", "Churn"]
df.drop(columns=cols_to_drop, inplace=True, errors="ignore")
```

```python
#Suppression des doublons
df.drop_duplicates(inplace=True)
```

```python
#Conversion des dates
df["Purchase Date"] = pd.to_datetime(df["Purchase Date"], errors="coerce")
```

```python
# Vérifications
print("\nValeurs manquantes après nettoyage :")
print(df.isnull().sum())

print("\nAperçu final :")
display(df.head())


#Sauvegarde
df.to_csv("data/ecommerce_clean.csv", index=False)
print(" Fichier nettoyé sauvegardé : data/ecommerce_clean.csv")
```
