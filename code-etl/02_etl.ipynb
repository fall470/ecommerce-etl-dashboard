```python
import pandas as pd

# Charger le CSV nettoyé
df = pd.read_csv("data/ecommerce_clean.csv", encoding="latin1")

# ---------------------------
#  Analyses et transformations
# ---------------------------


customers = df.groupby("Customer ID").agg(
    Age=("Customer Age", "first"),
    Gender=("Gender", "first"),
    Total_Purchases_Amount=("Total Purchase Amount", "sum"),
    Total_Transactions=("Purchase Date", "count"),
    Total_Quantity=("Quantity", "sum"),
    Total_Returns=("Returns", "sum")
).reset_index()

# Segmentation client
def segment_client(amount):
    if amount <= 10000:
        return "Low-Value"
    elif amount <= 15000:
        return "Mid-Value"
    else:
        return "High-Value"

customers["Segment"] = customers["Total_Purchases_Amount"].apply(segment_client)

# Customer Lifetime Value (CLV)
customers["CLV"] = (customers["Total_Purchases_Amount"] / customers["Total_Transactions"]).round(2)

# b) Product Category Analysis
product_analysis = df.groupby("Product Category").agg(
    Total_Revenue=("Total Purchase Amount", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Total_Transactions=("Purchase Date", "count"),
    Returns=("Returns", "sum")
).reset_index()

# c) Returns Analysis
returns_analysis = df.groupby("Customer ID")["Returns"].sum().reset_index()

# d) Payment Method Analysis
payment_method_analysis = df.groupby("Payment Method").agg(
    Total_Revenue=("Total Purchase Amount", "sum"),
    Total_Transactions=("Purchase Date", "count")
).reset_index()

# ---------------------------
# Sauvegarde dans des CSV
# ---------------------------

print("Sauvegarde dans des fichiers CSV...")

#df.to_csv("exports/fact_orders.csv", index=False, encoding="utf-8")
customers.to_csv("exports/customer_segmentation.csv", index=False, encoding="utf-8")
#product_analysis.to_csv("exports/product_category_analysis.csv", index=False, encoding="utf-8")
#returns_analysis.to_csv("exports/returns_analysis.csv", index=False, encoding="utf-8")
#payment_method_analysis.to_csv("exports/payment_method_analysis.csv", index=False, encoding="utf-8")

print("✅ Données sauvegardées en CSV avec succès (dossier 'exports/').")
```
